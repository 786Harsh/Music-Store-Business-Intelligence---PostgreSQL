### Q41 - Rank Tracks by Sales Within Each Genre
##### Business Question: Which tracks are the best-selling tracks within each genre?

```python
q41 = """
WITH track_sales AS (
    SELECT
        g.genre_id,
        g.name AS genre_name,
        t.track_id,
        t.name AS track_name,
        SUM(il.quantity) AS units_sold,
        ROUND(SUM(il.quantity * il.unit_price), 2) AS revenue
    FROM genre g
    JOIN track t
        ON g.genre_id = t.genre_id
    JOIN invoice_line il
        ON t.track_id = il.track_id
    GROUP BY
        g.genre_id,
        g.name,
        t.track_id,
        t.name
)

SELECT
    genre_name,
    track_id,
    track_name,
    units_sold,
    revenue,
    RANK() OVER (
        PARTITION BY genre_id
        ORDER BY units_sold DESC
    ) AS sales_rank
FROM track_sales
ORDER BY
    genre_name,
    sales_rank;
"""

df_q41 = pd.read_sql_query(q41, engine)

df_q41
```

### Q42 - Running Total of Revenue
##### Business Question: How does total revenue accumulate month by month?

```python
q42 = """
WITH monthly_sales AS (
    SELECT
        DATE_TRUNC('month', invoice_date) AS sales_month,
        ROUND(SUM(total), 2) AS monthly_revenue
    FROM invoice
    GROUP BY
        DATE_TRUNC('month', invoice_date)
)

SELECT
    sales_month,
    monthly_revenue,
    ROUND(
        SUM(monthly_revenue) OVER (
            ORDER BY sales_month
            ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
        ),
        2
    ) AS running_revenue
FROM monthly_sales
ORDER BY
    sales_month;
"""

df_q42 = pd.read_sql_query(q42, engine)

df_q42
```

### Q43 - Three-Month Moving Average
##### Business Question: What is the average monthly revenue over the current month and the previous two months?

```python
q43 = """
WITH monthly_sales AS (
    SELECT
        DATE_TRUNC('month', invoice_date) AS sales_month,
        ROUND(SUM(total), 2) AS monthly_revenue
    FROM invoice
    GROUP BY
        DATE_TRUNC('month', invoice_date)
)

SELECT
    sales_month,
    monthly_revenue,
    ROUND(
        AVG(monthly_revenue) OVER (
            ORDER BY sales_month
            ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
        ),
        2
    ) AS three_month_moving_average
FROM monthly_sales
ORDER BY
    sales_month;
"""

df_q43 = pd.read_sql_query(q43, engine)

df_q43
```

### Q44 - Top 3 Tracks in Each Genre
##### Business Question: What are the top 3 best-selling tracks in each genre?

```python
q44 = """
WITH track_sales AS (
    SELECT
        g.genre_id,
        g.name AS genre_name,
        t.track_id,
        t.name AS track_name,
        SUM(il.quantity) AS units_sold,
        ROUND(SUM(il.quantity * il.unit_price), 2) AS revenue
    FROM genre g
    JOIN track t
        ON g.genre_id = t.genre_id
    JOIN invoice_line il
        ON t.track_id = il.track_id
    GROUP BY
        g.genre_id,
        g.name,
        t.track_id,
        t.name
),

ranked_tracks AS (
    SELECT
        genre_name,
        track_id,
        track_name,
        units_sold,
        revenue,
        ROW_NUMBER() OVER (
            PARTITION BY genre_id
            ORDER BY units_sold DESC, track_id
        ) AS track_rank
    FROM track_sales
)

SELECT
    genre_name,
    track_id,
    track_name,
    units_sold,
    revenue,
    track_rank
FROM ranked_tracks
WHERE track_rank <= 3
ORDER BY
    genre_name,
    track_rank;
"""

df_q44 = pd.read_sql_query(q44, engine)

df_q44
```

##### Business Question: What are the main sales and customer performance metrics for the Chinook store?

```python
q45 = """
WITH sales_summary AS (
    SELECT
        COUNT(DISTINCT i.invoice_id) AS total_invoices,
        COUNT(DISTINCT i.customer_id) AS active_customers,
        ROUND(SUM(i.total), 2) AS total_revenue,
        ROUND(AVG(i.total), 2) AS average_invoice_value
    FROM invoice i
),

track_summary AS (
    SELECT
        SUM(il.quantity) AS total_units_sold,
        COUNT(DISTINCT il.track_id) AS different_tracks_sold
    FROM invoice_line il
),

top_track AS (
    SELECT
        t.name AS top_track_name,
        SUM(il.quantity) AS units_sold
    FROM track t
    JOIN invoice_line il
        ON t.track_id = il.track_id
    GROUP BY
        t.track_id,
        t.name
    ORDER BY
        units_sold DESC
    LIMIT 1
),

top_genre AS (
    SELECT
        g.name AS top_genre_name,
        SUM(il.quantity) AS units_sold
    FROM genre g
    JOIN track t
        ON g.genre_id = t.genre_id
    JOIN invoice_line il
        ON t.track_id = il.track_id
    GROUP BY
        g.genre_id,
        g.name
    ORDER BY
        units_sold DESC
    LIMIT 1
)

SELECT
    s.total_invoices,
    s.active_customers,
    s.total_revenue,
    s.average_invoice_value,
    ts.total_units_sold,
    ts.different_tracks_sold,
    tt.top_track_name,
    tt.units_sold AS top_track_units_sold,
    tg.top_genre_name,
    tg.units_sold AS top_genre_units_sold
FROM sales_summary s
CROSS JOIN track_summary ts
CROSS JOIN top_track tt
CROSS JOIN top_genre tg;
"""

df_q45 = pd.read_sql_query(q45, engine)

df_q45
```
